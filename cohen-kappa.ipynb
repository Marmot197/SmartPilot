```python
import pandas as pd
```

```python
df= pd.read_csv("c3an evaluation final v2 - Sheet13.csv")
df
```

```python
from sklearn.metrics import cohen_kappa_score

kappa = cohen_kappa_score(df['Groq Score'], df['Openai Score'], weights='quadratic')
print(f"Openai and Groq: {kappa:.4f}")
```

```python
from sklearn.metrics import cohen_kappa_score

# List of all raters
raters = ['Openai Score', 'Groq Score', 'Irfan', 'Fadi', 'Chathurangi', 'Aryaman','Moji']

# Prepare an empty DataFrame to store kappa scores
kappa_matrix = pd.DataFrame(index=raters, columns=raters, dtype=float)

# Compute Cohen's kappa between every pair of raters
for r1 in raters:
    for r2 in raters:
        if r1 == r2:
            kappa_matrix.loc[r1, r2] = 1.0  # Perfect agreement with self
        else:
            kappa = round(cohen_kappa_score(df[r1], df[r2], weights='quadratic'),3)
            kappa_matrix.loc[r1, r2] = kappa

display(kappa_matrix)
kappa_matrix.to_clipboard()
```
